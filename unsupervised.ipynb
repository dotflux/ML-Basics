```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs # generates datasets
from sklearn.cluster import KMeans, DBSCAN
from sklearn.preprocessing import StandardScaler # scaling is important in distancee based algorithms
```

```python
X, y_true = make_blobs(n_samples=500, centers=3, cluster_std=0.60, random_state=42)
```

```python
X
```

```python
df = pd.DataFrame(X, columns=['Feature_1','Feature_2'])
```

```python
df
```

# Scaling

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
```

```python
X_scaled
```

# K-Means Clustering

```python
wcss = [] # also known as inertia
K_range = range(1,11)
```

```python
for k in K_range:
    kmeans = KMeans(n_clusters=k,random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
wcss
```

```python
plt.plot(K_range,wcss,marker='o')
```

```python
kmeans_final = KMeans(n_clusters=3,random_state=42)
```

```python
cluster_labels = kmeans_final.fit_predict(X_scaled)
```

```python
cluster_labels
```

```python
df['cluster'] = cluster_labels
```

```python
sns.scatterplot(x = df['Feature_1'], y = df['Feature_2'], hue=df['cluster'], palette='viridis')
```

# K-Means in Non-Circular data

```python
from sklearn.datasets import make_moons
```

```python
X, y_true = make_moons(n_samples=100, noise=0.05, random_state=42)
```

```python
df = pd.DataFrame(X, columns=['Feature_1','Feature_2'])
```

```python
df
```

```python
X_scaled = scaler.fit_transform(df)
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans_labels = kmeans.fit_predict(X_scaled)
```

```python
df['kmeans_cluster'] = kmeans_labels
```

```python
sns.scatterplot(x = df['Feature_1'], y = df['Feature_2'], hue=df['kmeans_cluster'], palette='viridis')
```

# DBSCAN

```python
# had to test for perfect value of eps

for eps in [0.3, 0.35, 0.4, 0.45, 0.5]:
    dbscan = DBSCAN(eps=eps, min_samples=5)
    labels = dbscan.fit_predict(X_scaled)

    print(eps, len(set(labels)) - (1 if -1 in labels else 0))

for eps in [0.45, 0.5]:
    dbscan = DBSCAN(eps=eps, min_samples=5)
    labels = dbscan.fit_predict(X_scaled)

    print(f"eps: {eps}")
    print(f"Clusters: {len(set(labels) - {-1})}")
    print(f"Noise points: {sum(labels == -1)}")

    sns.scatterplot(
        x=df['Feature_1'],
        y=df['Feature_2'],
        hue=labels,
        palette='viridis'
    )
    plt.title(f'DBSCAN with eps={eps}')
    plt.show()
```

```python
dbscan = DBSCAN(eps=0.45, min_samples=5)
dbscan_labels = dbscan.fit_predict(X_scaled)
```

```python
df['dbscan_cluster'] = dbscan_labels
```

```python
sns.scatterplot(x = df['Feature_1'], y = df['Feature_2'], hue=df['dbscan_cluster'], palette='viridis')
```
